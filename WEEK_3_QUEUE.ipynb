```python

from collections import deque

ticket_line = deque()


def join_line(queue, name):
    queue.append(name)


def serve_customer(queue):
    if len(queue) == 0:
        print("No customers waiting.")
    else:
        return queue.popleft()


def show_line(queue):
    if len(queue) == 0:
        print("No customers waiting.")
    else:
        for i, name in enumerate(queue, start=1):
            print(i, name)


join_line(ticket_line, "Beatrice")
join_line(ticket_line, "Brian")
join_line(ticket_line, "Mercy")
join_line(ticket_line, "Kevin")
join_line(ticket_line, "Faith")


print("Served:", serve_customer(ticket_line))
print("Served:", serve_customer(ticket_line))


print("\nRemaining customers:")
show_line(ticket_line)


join_line(ticket_line, "David")
join_line(ticket_line, "Ann")


print("\nServing remaining customers:")

while len(ticket_line) > 0:
    print("Served:", serve_customer(ticket_line))

```

```python
def enqueue(queue, item):
  queue.append (item)

def dequeue(queue):
  queue.pop (item)

def peek(queue):
  if is_empty(queue):
    print("Queue is empty")
  else:
    return queue[0]

def is_empty(queue):
  return len(queue)==0

def size(queue):
  return len(queue)

```
